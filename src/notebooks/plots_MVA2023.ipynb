```python
import matplotlib.pyplot as plt
import matplotlib.path as mpath
import numpy as np


star = mpath.Path.unit_regular_star(6)
circle = mpath.Path.unit_circle()
# concatenate the circle with an internal cutout of the star
verts = np.concatenate([circle.vertices, star.vertices[::-1, ...]])
codes = np.concatenate([circle.codes, star.codes])
cut_star = mpath.Path(verts, codes)
```

```python
import matplotlib.pyplot as plt
fig = plt.figure(figsize=[5,3])
ax = fig.add_subplot(1,2,1)

y_base = [29.7, 30.8, 32.1]
x = [10, 20, 50]
ax.plot(x, y_base, '--g', marker=circle, markersize=10)
y = [30.2, 31.9, 32.2]
ax.plot(x, y, '-b', marker=circle, markersize=10)
ax.title.set_text('DALES dataset')
plt.xlabel('Pct. of labeled data (%)')
plt.ylabel('mIoU (%)')
plt.grid(True)
ax.legend(['Scratch', 'Pre-trained'],loc='lower right')


ax = fig.add_subplot(1,2,2)

y_base = [51.4, 53.2, 55.9]
x = [12, 20, 50]
ax.plot(x, y_base, '--g', marker=circle, markersize=10)
y = [54.5, 55, 56.2]
ax.plot(x, y, '-b', marker=circle, markersize=10)
ax.title.set_text('LiDAR-CAT3 dataset')


ax.legend(['Scratch', 'Pre-trained'], loc='lower right')
plt.grid(True)
plt.xlabel('Pct. of labeled data (%)')
#plt.ylabel('mIoU (%)')
plt.show()

fig.set_dpi(100)
fig.tight_layout(pad=0.4)
# fig.subplots_adjust(right=0.85)
#plt.gcf()
fig.savefig('comparison_labeled_data' + '.png', dpi=200) 
```
